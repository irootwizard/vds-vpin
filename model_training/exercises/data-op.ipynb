```python
# 数据预处理：更多行和列 + 删除缺失值最多的列 + 转张量
import os
import pandas as pd
import torch

os.makedirs(os.path.join('..', 'data'), exist_ok=True)
data_file = os.path.join('..', 'data', 'house_tiny.csv')

# 创建包含更多行和列的原始数据集
with open(data_file, 'w') as f:
    f.write('NumRooms,Alley,LotArea,YearBuilt,Condition,Price\n')
    f.write('NA,Pave,8450,2003,Normal,208500\n')
    f.write('2,NA,9600,1976,Normal,181500\n')
    f.write('4,NA,11250,2001,Abnorml,223500\n')
    f.write('3,NA,9550,1915,Normal,140000\n')
    f.write('NA,NA,14260,2000,Normal,250000\n')
    f.write('3,Grvl,14115,1993,Abnorml,143000\n')
    f.write('2,NA,10084,2004,Normal,307000\n')

# 读取原始数据
raw_data = pd.read_csv(data_file)
print('原始数据集:')
print(raw_data)

# 按列序号划分特征和标签（最后一列是 Price）
inputs = raw_data.iloc[:, :-1].copy()
outputs = raw_data.iloc[:, -1].copy()

# 删除缺失值最多的列
missing_counts = inputs.isna().sum()
missing_counts_row=inputs.isna().sum(axis=1)
print(missing_counts)
print(missing_counts_row)
most_missing_col = missing_counts.idxmax()
most_missing_row=missing_counts_row.idxmax()
inputs = inputs.drop(columns=[most_missing_col])
print(f'\n删除缺失值最多的列: {most_missing_col}')
print(f'\n删除缺失值最多的行: {most_missing_row}')

# 数值列用均值填充，类别列用众数填充
num_cols = inputs.select_dtypes(include='number').columns
cat_cols = inputs.select_dtypes(exclude='number').columns

inputs[num_cols] = inputs[num_cols].fillna(inputs[num_cols].mean())
for c in cat_cols:
    inputs[c] = inputs[c].fillna(inputs[c].mode()[0])

# one-hot 编码
inputs = pd.get_dummies(inputs, dummy_na=False, dtype=int)
print('\n预处理后特征:')
print(inputs)

# 转换为张量格式
tensor_X = torch.tensor(inputs.to_numpy(dtype=float), dtype=torch.float32)
tensor_y = torch.tensor(outputs.to_numpy(dtype=float), dtype=torch.float32)
```
